# Creating a calculator session

when starting a calculator session in a notebook, we want to specify the filename of the session, and the resulting created equation will be stored in the current working directory, in an expressions subfolder.

On this canvas, you can quickly and effeicently manually simplify projectors.

The open canvas has two modes:

### Create mode

Here we can create a sum of projectors. The controlling commands are:
- Add a new line: click one of the + icons under the bottom-most line
- Add a symmetriser (S): click the symmetriser button, or double-click between two lines
- Add an antisymmetriser (A): click the antisymmetriser button, or double-click on an existing symmetriser
- Move a line: click and drag the line from it's current node to a new position.
- Delete any part of a projector: right-click it
- Delete a projector: right-click just to the left of it (on top of its prefactor)
- Add a new projector: click + or -, or press + or - on your keyboard
- Multiply a projector: pressing ctrl on your keyboard exposes the multiply button in the centre of the projector. Clicking it opens a fraction editor where you can toggle between numerator and denominator using the / button. Pressing enter or clicking outside the fraction editor multiplies that projector by your specified fraction.

### Evaluate mode

Here we can evaluate the created sum of projectors. The following controls are enabled in this mode:
- Expand an S or A: double-click on the S or A
- Use the recursive expansion identity on S/A: mouse-over the S or A top or bottom line and click on the exposed tear icon.
- Permute lines: we can move S/A or free lines up or down, but they are locked so that only equivalent projectors can be generated. Permuting lines may switch the sign of the overall projector due to the antisymmetriser property.
- Take the trace: toggle the trace button on/off to display the polynomial N-dependent trace of the projector sum. This is not idiot-proof; if you have a sum of a projector on 3 lines and a projector on 4 lines, the polynomial will sum the two traces as though the first was traced on three and the second on four lines. Therefore, when summing projectors, it's up to you to ensure they are defined on the same number of lines.
- Undo the last operation: press and hold ctrl to expose the undo button on each individual projector.
- Undo a new equation line: click the top-right undo button.
- Move terms left/right: click in the empty space of one projector and drag it left/right.

### Save equation state

the constructed equation can be saved by clicking 'save' in the canvas. We can re-load it in code. When we operate on an expression loaded in this way, we're operating on the last saved line of the equation (see more below).

```python
filename = 'test'
bt.create(session=filename)
```

## What's the point of simplifying manually?

Manual simplification reduces the computer's job, e.g., when taking the trace, or when multiplying projectors. Let's time the performance of a partial manual expansion vs computing with simplified expressions:

```python
p1_initial = bt.load('P1')

bt.create(session='P1')
```

```python
p1_expanded = bt.load('P1-expanded')
bt.create(session='P1-expanded')
```

```python
%timeit p1_initial.collapse().trace()
```

```python
%timeit p1_expanded.collapse().trace()
```

```python
p1_initial.collapse().trace() == p1_expanded.collapse().trace()
```

```python
%timeit p1_initial * p1_initial
```

```python
%timeit p1_expanded * p1_expanded
```

#### Caution when checking equality of projectors: 

if they don't immediately look the same, we don't (yet?) perform any fancy equality checking, even though they may be identical:

```python
p1_initial == p1_expanded
```

However, if we need to check this, collapsing each term into a permutation sum is the way to go. This is not the default equality check because the collapse operation can take a long time, depending on the size of the projector. It will be implemented automatically after I have a working fast automatic solver.

```python
p1_initial.collapse() == p1_expanded.collapse()
```

## Operations on projectors

Projectors can be added, multiplied, normalised and traced, with mostly standard operations:

```python
multiple = p1_expanded * p1_expanded # not yet canvas-displayable, but can be collapsed and traced

multiple.collapse().trace()
```

```python
sum = p1_expanded + 4 * p1_expanded / 3
sum.collapse().trace()
```
